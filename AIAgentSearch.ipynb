AI agents are capable of impressive reasoning, but they cannot operate reliably without access to real-time information. This unit explains the fundamental limitations of LLMs, why these limitations matter in practical scenarios, and how web search becomes a critical component in building dependable agents.

#### LLMs Have a Knowledge Cutoff
Every LLM is trained on static data. Once training ends, the model cannot learn new events or updates unless developers perform a new training cycle or fine-tune it.

This creates real problems for agents:

- News and global events shift daily.
- Company information, product details, and prices change constantly.
- Company information, product details, and prices change constantly.
- Technical documentation gets updated frequently.
- Scientific research evolves faster than training cycles.

For any task that depends on the present, the model's internal knowledge becomes stale almost immediately.

Example: Ask an LLM trained in early 2024 about API changes made in 2025. It will attempt to answer, but the content will be incomplete or wrong because those updates do not exist in its training data.

Search gives the agent access to information that the LLM has never seen.

##### Internal Reasoning Alone Causes Hallucinations
LLMs generate text by predicting likely sequences of tokens. Without verifiable external data, they often "invent" details that sound plausible but are factually incorrect.

Common hallucination patterns include:

- Citing articles that do not exist
- Creating fake statistics
- Mixing unrelated facts
- Confidently presenting outdated or wrong information
- Overgeneralizing from limited context

Web search grounds the agent in real sources, reducing the chance of incorrect or fabricated outputs.

Context Windows Are Limited
LLMs cannot store or reason over everything at once. Even the largest context windows cannot hold:

• Multiple long articles

• Detailed blog posts

• Full documentation sets

• Entire website sections

Agents must retrieve context dynamically rather than rely on preloaded information.

Search lets them:

Agents depend heavily on how they retrieve information. Understanding the three main types of search: keyword, semantic, and hybrid - helps you design smarter queries, reduce irrelevant results, and improve the accuracy of your agent's reasoning. This unit walks you through each search type, how they work, what their strengths are, and where they fall short.

Keyword Search
Keyword search is the simplest and oldest form of search. It looks for exact matches between the words a user types and the words found in documents.

How it Works
• Matches literal tokens in a query against web pages

• Works best when you know the exact words or phrases you're looking for

• Rankings rely on frequency, titles, headings, metadata, and URL patterns

Strengths
• Fast and efficient

• Great when the keywords are unique or specific

• Useful for highly factual or structured searches

• Example: "Python 3.12 release date"

• Example: Searching for code snippets with exact API names

Limitations
• Misses content that uses different phrasing

• Fails when intent doesn't match literal wording

• Struggles with vague or conceptual questions

• Sensitive to synonyms, typos, and variations

Example:

Query: "best way to reduce website bounce rate" - A keyword search may miss pages that use "decrease exit rate" or "improve engagement."

Agents relying only on keyword search often retrieve narrow, incomplete, or noisy results.

Semantic Search
Semantic search retrieves content based on meaning, not exact wording. Instead of matching characters, it compares the conceptual intent of the query with the intent of documents.

How it Works
• Uses embeddings to represent meaning

• Measures similarity between query vector and document vectors

• Finds content that is contextually or conceptually related

Strengths
• Understands synonyms and paraphrasing

• Handles complex or conversational queries

• Excellent for research tasks, explanations, comparisons, and broad topics

• Returns more relevant results even when exact words are absent

Examples:

Query: "how to increase user retention" - semantic search can find: "improve customer loyalty", "keep users coming back", "reduce churn in SaaS products"

Query: "AI tools that browse the web" - semantic search can surface articles describing agents or browser automation without using that exact wording.

Limitations
• May bring in broader or loosely related content

• Requires more computing power

• Quality depends heavily on the embedding model

Semantic search is the foundation of modern agent search because agents often ask open-ended, contextual questions.

Hybrid Search
Hybrid search combines keyword and semantic search to deliver the best of both worlds.

How it Works
• Performs semantic matching to understand intent

• Uses keyword relevance to anchor results to factual grounding

• Ranks results using a weighted combination of both signals

Strengths
• More accurate than using either method alone

• Great for queries that need both intent and precision

• Reduces hallucination risk by grounding conceptual matches in real keywords

• Particularly effective for multi-step agent workflows

Example:

Query: "tavily extract use cases"

Hybrid search can:

• Understand that "content extraction", "full-page parsing", and "clean HTML extraction" are conceptually related

• Prioritize pages that actually contain the phrase "Tavily Extract" or "extract API"

Limitations
• More expensive than using a single method

• Requires thoughtful tuning to balance semantic vs keyword weighting

• Can still surface noise if the query is too broad

When Agents Use Each Type
Understanding when to use each method helps agents work more efficiently.

Scenario	Best Search Type	Reason
You need high precision for very specific terms	Keyword	Exact matching
You want conceptually relevant results for broad questions	Semantic	Understands meaning
You need both precision and strong intent matching	Hybrid	Balanced approach
Research tasks needing deep exploration	Hybrid or Semantic	Captures broader context
Technical or documentation-based questions	Keyword + Semantic	Finds exact API names plus explanations
Why This Matters for Agents
Agents need search to be:

• Accurate: Retrieve exactly what the user meant

• Flexible: Understand synonyms and intent

• Grounded: Avoid hallucinations

• Resilient: Adapt to vague or incomplete queries

This combination allows agents to interpret human queries the way humans interpret information.

High-quality search means retrieving information that is complete, relevant, fresh, and trustworthy so the agent can reason accurately. When search quality drops - missing data, outdated sources, or noisy results - agent outputs degrade immediately.

Recall
Recall measures how many relevant documents are retrieved, even if some irrelevant ones appear too.

Precision
Precision measures how relevant the retrieved results are, even if the agent misses a few items.

Balancing both:

Agents often begin with high recall (broad search) and refine toward high precision (filtered search).

Recency
Recency determines how up-to-date your information is.

For agents that deal with fast moving topics (for example news, prices, policies, and more..), stale information can lead to incorrect conclusions.

Domain Filters
Domain filters let agents control which sources they use, while domain trust scoring evaluates how credible each site is. By including or excluding specific domains, agents focus on authoritative sources and avoid wasting time or tokens on low-quality content.

Depth and Exploration
Depth determines how far a search engine explores past the initial set of results.

When a task is complex, deep searches reveal more context and higher precision.

Query Refinement and Expansion
Agents rarely get the perfect answer from a single query. They must refine or expand their search based on what they find.

Failure Modes
Agents often struggle due to predictable search failures. Recognizing these helps you design better agent workflows. Some examples of failure modes:

Too broad query. Fix: Add filters, specify topic, increase precision.

Too narrow query. Fix: Expand the query, generalize, or increase depth.

Outdated Sources. Fix: Apply recency filters or validate timestamps.

Low-Quality Domains. Fix: Exclude domains or prioritize trustworthy sources.

Why These Concepts Matter for Agents
Search quality directly impacts the agent's ability to:

• Ground answers in reality

• Generate accurate summaries

• Avoid hallucinations

• Perform deep research

• Produce consistent outputs across runs

Good search is not just about retrieving content, it's about retrieving the right content.
